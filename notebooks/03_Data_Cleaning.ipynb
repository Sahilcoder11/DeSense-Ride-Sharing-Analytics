```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("DeSense_Ride_Sharing_Messy_Dataset.csv")
```

```python
df.isnull().sum()
```

```python
df['Fare_Amount']=df['Fare_Amount'].fillna(df['Fare_Amount'].median())

df['Driver_Rating']=df['Driver_Rating'].fillna(df['Driver_Rating'].median())

df['Driver_ID']=df['Driver_ID'].fillna(df['Driver_ID'].mode()[0])

df['Pickup_Location']=df['Pickup_Location'].fillna(df['Pickup_Location'].mode()[0])

df['Dropoff_Location']=df['Dropoff_Location'].fillna(df['Dropoff_Location'].mode()[0])

df['Payment_Method']=df['Payment_Method'].fillna(df['Payment_Method'].mode()[0])
```

```python
df.isnull().sum()
```

**Pickup_Time and Dropoff_Time:**

Missing values in these columns were not filled using mode because time values are related to individual ride schedules. Replacing them with the most frequent time could create inaccurate records and affect time-based analysis. Therefore, these missing values were kept for further investigation.

**Convert Ride_Date into datetime format.

```python
df['Ride_Date'] = pd.to_datetime(
    df['Ride_Date'], format='mixed', errors='coerce', dayfirst=True
)
```

```python
print(df['Ride_Date'].dtype)
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = df[col].str.strip()
```
