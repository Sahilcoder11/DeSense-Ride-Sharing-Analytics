```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("DeSense_Ride_Sharing_Cleaned_Dataset.csv")
```

```python
df['Ride_Date'] = pd.to_datetime(df['Ride_Date'], errors='coerce')
```

```python
print(df['Ride_Date'].dtype)
```

```python
df['Ride_Year'] = df['Ride_Date'].dt.year
print(df['Ride_Year'])
```

```python
df['Ride_Month'] = df['Ride_Date'].dt.month
```

```python
df['Fare_Per_KM'] = (
    df['Fare_Amount'] / df['Ride_Distance_km']
)
```

```python
df['Ride_Status_Category'] = np.where(
    df['Ride_Status'] == 'Completed',
    'Successful',
    'Not Completed'
)
```

```python
df[['Ride_Date', 'Ride_Year', 'Ride_Month',
    'Fare_Per_KM', 'Ride_Status_Category']].head()
```

calculating ride duration time

```python
df['Pickup_Time'] = pd.to_datetime(
    df['Pickup_Time'], format='%H:%M', errors='coerce'
)

df['Dropoff_Time'] = pd.to_datetime(
    df['Dropoff_Time'], format='%H:%M', errors='coerce'
)
```

```python
df['Ride_Duration_Minutes'] = (
    df['Dropoff_Time'] - df['Pickup_Time']
).dt.total_seconds() / 60

df.loc[
    df['Ride_Duration_Minutes'] < 0,
    'Ride_Duration_Minutes'
] += 1440
```
